```python
import json
from pathlib import Path

compounds_path = Path("EDA_Names.json")
drugs_path = Path("egyptian-drugs.json")

# Load compounds
with compounds_path.open("r", encoding="utf-8") as f:
    compounds = json.load(f)

# Load drugs
with drugs_path.open("r", encoding="utf-8") as f:
    drugs = json.load(f)

# Get all scientific names from drugs
scientific_names = {
    drug.get("scientific_name", "").strip().upper()
    for drug in drugs
}

# Remove compounds that don't exist in drugs
compounds = [
    compound
    for compound in compounds
    if compound["compound_name"].strip().upper() in scientific_names
]

# Save the filtered compounds
with compounds_path.open("w", encoding="utf-8") as f:
    json.dump(compounds, f, indent=4, ensure_ascii=False)

print(f"Remaining compounds: {len(compounds)}")
```
